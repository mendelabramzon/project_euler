# Problem 2

```python
sum=0
phi = (1 + 5**0.5) / 2
mu = (1 - 5**0.5) / 2
for i in range(1, 4000000):
    fib = (phi**i - mu**i) / 5**0.5
    if fib > 4000000:
        break
    if fib % 2 < 1:
        sum += fib
print(sum)


sum2 = 0
f0 = 1
f1 = 1
while f1 < 4000000:
    f0, f1 = f1, f0 + f1
    if f1 % 2 == 0:
        sum2 += f1
print(sum2)

```
